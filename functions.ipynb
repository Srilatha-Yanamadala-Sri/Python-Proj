```python
def greet():

    print("Hej! Welcome")

greet()

def name():
    name = input("What is your name? ")
    print(f"Nice to meet you, {name}!")

name()

def greet_person(first_name = "Jane", last_name = "doe"):
    print(f"Hello, {first_name} {last_name}!")

greet_person(last_name = "Smith", first_name = "Jane")
```

```python
def calculate_area(length, width):
    area = length * width
    print(f"The area of the rectangle is: {area}")

calculate_area(10, 5)
 
```
