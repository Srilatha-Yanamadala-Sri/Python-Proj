```python
age = int(input("Enter your age: "))

if age == 20:
    print(" you are equal to 20")

elif age < 20:
    print(" you are less than 20")

elif age > 20:
    print(" you are greater than 20")

else:
    print(" you are not in this if condition")
```

```python
print(age == 20)
print(age != 20)
print(age < 20)
print(age > 20)
```

```python
ready = input("Are you ready? (y/n): ")

if ready == "y":
    print("You are ready!")
elif ready == "n":
    print("You are not ready.") 
else:
    print("Invalid input. Please enter 'y' or 'n'.")
```
