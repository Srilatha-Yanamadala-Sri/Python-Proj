# Skriva Files


### When we try to write to a file that does not exist, it will be created automatically.
```

````python
with open("files/example.txt", "w", encoding = "utf-8") as f: # w is used to write text in the new/create file
    f.write("Hello world\n")
    f.write("This is the first line in the file\n")
````

````python
with open("files/example.txt", "a", encoding = "utf-8") as f: # a is used to add the text in the file
    f.write("This is the second line in the file\n")
````

````python
with open("files/text.txt", "a", encoding = "utf-8") as f: # a is used to add/create the text in the file
    f.write("This is the first text in the test file\n")
````

````python
lines = ["Rad 1\n", "Rad 2\n", "Rad 3\n"]

with open("files/multilines.txt", "w", encoding = "utf-8") as f: # w is used to write text in the new/create file
   f.writelines(lines)
````

````python
with open("files/test.txt", "w", encoding = "utf-8") as f: # w is used to write text in the new/create file
    print("Hello world\n", file=f)
    print("This is the first line in the file\n", file=f)
````

````python
with open("files/test.txt", "a", encoding = "utf-8") as f: # a is used to write/add text in the file
    print("Hello Srilatha\n", file=f)
    print("This is the second line in the file\n", file=f)
````

````python
while True:
    note = input(" write a sentence or type q for exit : ")
    if note.lower() == "q":
        break
    with open("files/notes.txt", "a", encoding = "utf-8") as f:
        f.write(note + "\n")
````
