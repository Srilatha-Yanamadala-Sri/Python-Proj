# Numbers

```python
x = 10      #int = heltal 
y = 0.03    #float = decimaltal

print(" x = ", type (x ))
print(" y = ", type (y)) 
```

```python
print(7 + 3)  # addition
print(7 - 3)  # subtraktion
print(7 * 3)  # multiplikation
print(7 / 3)  # division
print(7 // 3)  # heltalsdivision
print(7 % 3)   # modulus 
```

```python
x = 10
y = 3.4
print(float(x))
print(int(y))
```

```python
num = -7

print(abs(num))  # absolute value
print(pow(2, 3))  # exponentiation
print(round(3.14159, 2))  # rounding to 2 decimal places
print(max(1, 5, 3))  # maximum value
print(min(1, 5, 3))  # minimum value
```

```python
import math
print(math.sqrt(16))  # square root
print(math.sin(math.pi / 2))  # sine of 90 degrees
print(math.pi)  # value of pi
print(math.floor(3.14))  # floor of 3.14
print(math.ceil(3.14))  # ceiling of 3.14
```
