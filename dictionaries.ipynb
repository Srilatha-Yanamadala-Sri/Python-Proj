# Dictionaries


```python
person={
    "name" : "Srilatha",
    "age" : 37,
    "city" : "Helsingborg"
}

print(person)
print(person["name"])
print(person["age"])
```

```python
person["age"] = 38
person["email"] = "srilatha@gmail.com"
del person["city"]

print(person)
```

```python
print(person.keys())
print(list(person.keys()))
print(person.values())
print(person.items())
```

```python
for key, value in person.items():
    print(key,":", value)
```

# Sets

```python
fruits = ["apple", "banana", "orange", "apple", "kiwi", "banana"]
#fruits = {"apple", "banana", "orange", "kiwi"}

fruits_set = set(fruits)  # Convert to set to remove duplicates

print(fruits)

print(fruits_set)
```

```python

def cars_color_count(dictionary):
    colors = list(dictionary.values())


    for color in set(colors):
        num = colors.count(color)
        print(f"There are {num} {color} cars")

def cars_in_garage(dictionary):
    for key, val in dictionary.items():
        print(f"{key} is a {val} car")


cars = {}


while True:
    reg = input("Enter your registration number: ")
    color = input("Enter your car color: ")

    cars[reg] = color

    another = input("Do you want to add another car? (y/n): ")
    if another.lower() == "y":
        continue
    else:
        break

cars_in_garage(cars)
cars_color_count(cars)
```
