# Read/Write CSV files and JSON files

### CSV = comma seperated values

```python
import csv

with open("csv_files/cars.txt", "w", newline = "",  encoding = "utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["reg", "color"])
    writer.writerow(["ABC001", "Red"])
    writer.writerow(["BCd002", "Green"])
    writer.writerow(["CDE003", "Blue"])
```

```python
import csv

with open("csv_files/cars.txt", "r", encoding = "utf-8") as f:
    reader = csv.reader(f)
    for row in reader:
        print(row)

        print(row[0])
```

# JSON - Javascript obeject notation

```python
import json

data = {
    "name" : "Srilatha",
    "age" : 37,
    "city" : "Helsingborg",
    "is_marries" : True
}

with open("csv_files/person.json", "w", encoding = "utf-8") as f:
    json.dump(data, f, indent=4, ensure_ascii=False)
```

```python
import json

with open("csv_files/person.json", "r", encoding = "utf-8") as f:
    person = json.load(f)
    
print(person)
```
