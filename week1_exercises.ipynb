### 4. Strängar – indexering, slicing, sammanslagning
Utgå från `name = "Ada Lovelace"`.
1. Skriv ut antalet tecken, de tre första tecknen och de två sista tecknen i strängen.
2. Plocka ut förnamnet och efternamnet (hint: `split`).
3. Skriv `Hej, <förnamn>!` med f-string.
4. Skapa en ny sträng där mellanslag ersätts med `_`.
5. Skriv ut strängen baklänges med slicing.

```python
# Din kod här
name = "Ada Lovelace"

#1. Skriv ut antalet tecken, de tre första tecknen och de två sista tecknen i strängen.
print("Antal tecken:", len(name))
print("Första tre tecknen:", name[:3])
print("Sista två tecknen:", name[-2:])

#2. Plocka ut förnamnet och efternamnet (hint: `split`).
förnamn = name.split()[0]
efternamn = name.split()[1]
print("Förnamn:", förnamn)
print("Efternamn:", efternamn)

#3. Skriv `Hej, <förnamn>!` med f-string.
print(f"Hej, {förnamn}!")

#4. Skapa en ny sträng där mellanslag ersätts med `_`.
ny_sträng = name.replace(" ", "_")
print("Ny sträng:", ny_sträng)

#5. Skriv ut strängen baklänges med slicing.
print("Strängen baklänges:", name[::-1])


```

### 8. Listor – grunder

3. Skapa en ny lista med de tre första elementen.
4. Lägg till `10` sist i nums.
5. Räkna hur många gånger `2` förekommer.
6. Skapa en **sorterad kopia** utan att förstöra originalet.

```python
# Din kod här
nums = [5, 1, 7, 2, 9, 2]

#2. Skriv ut första och sista elementet.
print("First element:", nums[0])
print("Last element:", nums[-1])

#3. Skapa en ny lista med de tre första elementen.
tre_första = nums[:3]
print("The first three elements:", tre_första)

#4. Lägg till `10` sist i nums.
nums.append(10)
print("The list after adding 10:", nums)

#5. Räkna hur många gånger `2` förekommer.
antal_tvåor = nums.count(2)
print("Number of times '2' appears:", antal_tvåor)

#6. Skapa en **sorterad kopia** utan att förstöra originalet.
sorterad_kopia = sorted(nums)
print("Sorted copy:", sorterad_kopia)
print("Original list:", nums)  
```
