# Read files


```python
file = open("files/cities.txt", "r") #r = read
cities_content = file.read() # read the whole file here 
file.close() # very imp step. We have to close after read.

file = open("files/people.txt", "r") #r = read
people_content = file.read()
file.close()


cities = [line.strip() for line in cities_content.splitlines() if line.strip()]
people_counts = [line.strip() for line in people_content.splitlines() if line.strip()]

for city, count in zip(cities, people_counts):
    print(f"{city}: {count}")
```
