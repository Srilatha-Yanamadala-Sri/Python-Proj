# Strings

```python
str = " hello world"
print(str.strip())  # remove leading and trailing whitespace
print(str.upper())  # convert to uppercase
print(str.lower())  # convert to lowercase
print(str.replace("world", "Python"))  # replace a substring
print(str.split())  # split the string into a list of substrings
```

```python
first_name = "Alice"
last_name = "Smith"
text = f"My name is {first_name} {last_name}."  # f-string formatting
age = 30

print("Hello, my name is " + first_name + " " + last_name + ".")  # string concatenation
print("Hello, my name is {} {} and I am {} years old.".format(first_name, last_name, age))  # string formatting with format()

print(text)  # output: My name is Alice Smith.

```
